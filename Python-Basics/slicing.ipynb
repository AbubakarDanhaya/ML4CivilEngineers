### **`SLICING`**
Slicing in Python is a technique to extract a portion (subset) of a sequence such as a list, tuple, string, or array using index positions. We normally employ this technique when working with a specified groups of items in a list. To make a slice, you specify the index of the first and last elements you want to work with. 

```
for example

names = ["james", "daniel", "umar", "abdu", "ali"]
yoruba_names = names[0:2]

```

````python
names = ["james", "daniel", "umar", "abdu", "ali"]
yoruba_names = names[0:2]
print("The following are Yoroba names")
for name in yoruba_names:
    print(name.title())
````

````python
print("These are my Hausa friends\n", names[2:5])
````

````python
players =["charles","martina", "micheal", "florence", "eli"]
print("first three players are: ")
for player in players[0:3]:
    print(player.title())
````

**_If you omit the first index in a slice, Python starts at the begining of the list_.**

````python
print("First four players: ")
for player in players[:3]:
    print(player.capitalize())
````

**_A similar syntax works if you want a slice that includes the end of the list. for example, if you want all the items from the third item through the last item, you can start with index 2 and omit the second index_**

````python
print("name of players from the Third to last :")
for player in players[2:]:
    print(player.capitalize())
````

### **Using range() Function to make a List of Numbers**
If you want to make a list of numbers, you can convert the results of **range()** directly into a list using the **list()** function.

````python
numbers = list(range(10))
print(numbers)
````

**You can provide start and terminal value of items in the list by specifying them in range() function**

````python
nums = list(range(1, 11))
print(nums)
````

**Take note the last item is not included always**

we can also use the range function to tell the python to skip numbers in a given range. if you can pass a third argumentto range(), Python uses that value as a step size when generating numbers.

````python
even_nums = list(range(2, 11, 2))
print("These are even numbers:", even_nums)
````

You can create almost any set of numbers you want to using the range() function. lets build a list of squares.

````python
squares =[]
for num in range(1, 11):
    squares.append(num**2)
print("the following are squares of 1 to 10\n", squares)
````

````python
import numpy as np

nums_with_rt = []

for num in range(1, 101):
    if np.sqrt(num).is_integer():
        nums_with_rt.append(num)

print("The following are numbers with integer square roots:")
for num in nums_with_rt:
    print(num)
````

### **Simple Statistics with a List of Numbers**
A few Python functions are helpful when working with lists of numbers. For example, you can easily find the minimum, maximum and sum of a list of numbers.

````python
digits = list(range(1, 11))
print(digits)
````

let's filter minimum, maximum and find sum of these numbers in this list.

````python
print("minimum number in the list is ", min(digits))
print("Maximum number in the list is ", max(digits))
print("The sum of these numbers in the list is ", sum(digits))
````
