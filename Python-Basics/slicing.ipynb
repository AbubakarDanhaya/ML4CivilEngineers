# SLICING IN PYTHON PROGRAMMING

### `SLICING`
The technique called slicing is normally employ when working with a specified groups of items in a list. To make a slice, you specify the index of the first and last elements you want to work with. 

```
for example

names = ["james", "daniel", "umar", "abdu", "ali"]
yoruba_names = names[0:2]

```

````python
names = ["james", "daniel", "umar", "abdu", "ali"]
yoruba_names = names[0:2]
print("The following are Yoroba names")
for name in yoruba_names:
    print(name.title())
````

````python
print("These are my Hausa friends\n", names[2:5])
````

````python
players =["charles","martina", "micheal", "florence", "eli"]
print("first three players are: ")
for player in players[0:3]:
    print(player.title())
````

**_If you omit the first index in a slice, Python starts at the begining of the list_.**

````python
print("First four players: ")
for player in players[:3]:
    print(player.capitalize())
````

**_A similar syntax works if you want a slice that includes the end of the list. for example, if you want all the items from the third item through the last item, you can start with index 2 and omit the second index_**

````python
print("name of players from the Third to last :")
for player in players[2:]:
    print(player.capitalize())
````
