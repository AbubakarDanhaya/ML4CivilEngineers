### `What is a List`?

A `list` is a collection of items in a particular order. You can make a list that includes the letters of alphabet, the digit from 0-9, or the names of all the people in your family. You can put anyting you want into a list, and the items in your list don't have to be related in any particular way.

In Python, square brackets ([ ]) indicate a list , and individual elements in the list are separated by commas.

### `Accessing Elements in a List`
Lists are ordered collections, so you can access any element in a list by telling python the position, or index, of the itemdesired. To access an element in a list, write the name of the list followed by the index of the itemenclosed in square brackets.

```python
names = ["isah", "umar", "abba", "abubakar "]
print(names[3].strip().title())
```

let's print the second and third elements of the list.

```python
second_name = names[1].title()
third_name = names[2].title()

print("second name and third name are ", second_name + "and " + third_name + " respectively")
```

### `Index Postions Start at 0, Not 1`
Python considers the first item in a list to be at position 0, not position 1. This is true of most programming languages, and the reason has to do with how the list operations are implemented at a lower level. The second item in alist has index of 1.

**REMEMBER** : Python has special syntex for accessing the last element in alist. By asking for the item at index **-1**, python always returns the last item in the list.

```python
# Let's try psitive indexing
print("first name in the list is ", names[0].title())
print("second name is ", names[1].title())
print("last name in the list is ", names[3].title())
```

```python
# Negative indexing
print("the last name in the list is ", names[-1].title())
print("the first name in the list is ", names[-4].title())
```

### `Changing, Adding, and Removing Elements`
Most lists you created are dynamic, meaning you'll build a list and then add and remove elements from it as your programs runsits course. 

for example you can create an empty list can keep building as your program runs.

#### `Modifying Elements in a list`
The syntax for modifying an element is similar to the syntax for accessing an element in a list. To change an element, use the name of the list followed by the index of the element you want to change, and then provide the new value you want that item to have.

for example let assume we have a list of motorcycles that contains **honda, yamaha and suzuki**. pls remove the **honda** and replace it with **ducati**.

```python
motorcycles = ["honda", "yamaha", "suzuki"]
print("original motorcycles:", motorcycles)
motorcycles[0] = "ducati"
print("modified motorcycles:", motorcycles)
```

### `Adding Elements to a List`
You might want to add a new element to a list for many reasons. python provides several ways to add new data to existing lists.

#### `Appending Elements to the End of a List (.append())`
The simplest way to add a new element to a listis to _append_ the item to the list. When you append an item to a list, the new element is added to the end of the list. Using the same list we had in the previous example. 

**.append( )** methods add element at the end of the list

```python
names = ["isah", "umar", "abba", "abubakar"]
names.append("musa")
print(names)
```

```python
motorcycles = []
motorcycles.append("honda")
motorcycles.append("yamaha")
motorcycles.append("suzuki")
print(motorcycles)
```

#### `Insert Elements into a List (.insert(index, value))`
You can add a new element at any position in your list using the **.insert()** methods. You can do this by specifying the index of the element and and the vaue of the new item.

```python
courses = ["soil", "concrete", "steel", "geotech"]
# Let us insert hydrology at index 4
courses.insert(4, "hydrology")
print(courses)
```

```python
# Let us insert transportation at index 2
courses.insert(2, "transportation")
print(courses)
```

#### `Removing Elements from a List`
Often, you'll want to remove an item or set of items from a list. You can remove an item **_according to its position in the list_** or **_according to its value_**

```python
fruits = ["mango", "banana", "orange", "apple"]
print(fruits)
```

##### `Removing an item Using del Statement`

```python
del fruits[2]
print(fruits)
```

#### `Removing an item Using the .pop() Methods`
sometimes you'll want to use the value of an item after you remove it from a list. The **.pop( )** method removes the last item in a list, but it lets you work with that item after removing it.

```python
courses = ["math", "physics", "chemistry"]
removed_course = courses.pop()
print("The removed course is:", removed_course.title())
print(courses)
```

**NOTE**: You can also use **.pop( )** method with index to locate the specific element you want to remove.

```python
motorcycles = ["honda", "yamaha", "suzuki", "ducati", "bmw", "marcedes"]
poped_motorcycle = motorcycles.pop(3)
print("The poped motorcycle is:", poped_motorcycle.title())
print(motorcycles)
```

#### `Removing Item by Value`
Sometimes you won't know the position of the value you want to remove from a list. If you on;y know the value of the item you want to remove, you can use the **.remove( )** method.

```python
tv_shows = ["got", "breaking bad", "the wire", "friends", "the sopranos"]
print(tv_shows)
```

```python
tv_shows.remove("friends")
print(tv_shows)
```

## `Organising a List`
Python provides a number of different ways to organize your lists, depending on the situation.
- sorting permanetly with the **.sort()** method
- sorting temporarily with the **sorted()** function  

### `.sort() Method`

```python
social_media = ["facebook", "twitter", "instagram", "whatsapp", "telegram"]
print("social_media:", social_media)
social_media.sort()
print("Sorted social media:", social_media)
```

```python
social_media
```

```python
cars = ["bmw", "audi", "toyota", "subaru"]
print(cars)
```

```python
cars.sort()
print(cars)
```

you can sort in reverse alphabetic order by setting reverse = true in the sort() method.

e.g **.sort(reverse=True)**

```python
cars = ["bmw", "audi", "toyota", "subaru"]
print("Here is the original list:", cars)

cars.sort(reverse=True)
print("Here is the sorted list in reverse order:", cars)
```

```python
foods = ["pizza", "burger", "pasta", "salad", "sushi"]
print("Original foods list:", foods)
foods.reverse()
print("Reversed foods list:", foods)
```

#### `sorted() Function`

```python
branches_civil = ["structural", "geotech", "transportation", "water resources", "environmental"]
print("Original branches of civil engineering:", branches_civil)
print("\nHere is the sorted list temporarily:", sorted(branches_civil))
print("\nOriginal branches of civil engineering after using sorted():", branches_civil)
```

## `Length of a List`
You can find the length or number of items in a list using **len()** function.

```python
print("length of fruits list:", len(fruits))
print("The number of fruits available are:", len(fruits))
```
