# **LIST COMPREHENSIONS**
### **Definitions**
**List comprehension in Python is concise, readable and efficient way to create a new list by applying an expression to each item in an existing iterable(like list, tuple or a range), optionally filtering the items based on a condition. This technique is an alternative to a traditional `for loop` and `map()` and `filter()` functions for list creations.**

### **Syntax**
The basic syntax of a list comprehension is a sigle line of code enclosed in square brackets.
```
new_list = [expression for item in iterable if condition]
```
- expression: The value or operation applied to each item. This is the outcome that gets added to the new list.
- item:  A variable representing the current item in the iteration.
- iterable: The source sequence or collection (e.g, a list, string, tuple or range) being iterated over.
- if condition (optional): A filter that only accepts items where the condition evaluates to `True`.

**Creating list using Traditional  for loop method**

````python
squares = []
for num in range(1, 11):
    squares.append(num**2)
print(squares)
````

**Using list comprehensions**

````python
squares = [x**2 for x in range(1, 11)]
print(squares)
````

List comprehension can be used for : <br>
**1. Mapping:**<br> returns a new list that results from performing an operation on each element of a given list or range. E.g returns cubes of nums in a given range.

````python
cubes = [x**3 for x in range(1, 11)]
print(cubes)
````

````python
names = ["abba", "umar", "salim"]
````

**Using for loop methods**

````python
names_upper = []
for name in names:
    names_upper.append(name.upper())
print(names_upper)
````

**Using List comprehensions**

````python
names_upper = [name.upper() for name in names]
print(names_upper)
````

**2. Filtering:**<br>
Returns a new list that results from keeping those elements of a given list or range that satisfy condition(s). E.g return a list of names in people whose name start with a vowel.

````python
names = ["ali", "salim", "sani", "abba", "umar", "faruk"]
````

````python
names_st_vwl = [name for name in names if name[0] in ["a", "i", "u", "o", "e"]]
print(names_st_vwl)
````
