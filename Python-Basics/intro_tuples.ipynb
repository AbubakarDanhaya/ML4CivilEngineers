## **Advantages of Tuples Over Lists**
**_In Python, tuples and lists are both used to store ordered collections of items, but tuples have some important advantages over lists_**. 

**1. Immutability (Data Safety)**<br>
Tuples are immutable once created they cannot be changed. This makes tuple safe for the data that should not change (e.g dimensions, students details, coordinates, database records etc)

**2. Faster Performance** <br>
Tuples are generally faster than lists for iteration and access. the reason is python does not need to manage resizing or nodifications.

**3. Less Memory Usage** <br>
Tuples consume less memory than lists, this is useful when working with large datasets or memory-constrained systems.

**4. Can Be Used as Dictionary Keys** <br>
Tuples (If they contain immutable elements) are hashable while list are not hashable and cannot be dictionary keys.

```
location = {(10.5, 20.3): "Bridge A",(12.8, 11): "Bridge B"}
```

````python
nums = list(range(1, 1000))
print(nums)
print("The data type of nums is:", type(nums))
print("The space used by nums is:", nums.__sizeof__(), "bytes")
````

````python
nums = tuple(range(1, 1000))
print(nums)
print("The data type of nums is:", type(nums))
print("The space used by nums is:", nums.__sizeof__(), "bytes")
````

## **Looping Through All Values in a Tuple**
You can loop over all the values in a tuple using a for loop, just as you did with a list:

````python
dimensions = (2560, 1440)
for dimension in dimensions:
    print(dimension)
````

````python
student_details = ("Aisha", "Mechanical Engineering", 3.9, 2023)
print(f" Student name is {student_details[0]} and  she is in {student_details[1]} \
with a GPA of {student_details[2]} graduating in {student_details[3]}.")
````

### **Writing Over a Tuple**

**Although you can't modify a tuple, you can assign a new value to a variable that represent a tuple. for example, if we wanted to change the dimensions of the above rectange, we could redefine the entire tuple:**

````python
dimensions = (1280, 720)
print("Original dimensions:")
for dimension in dimensions:
    print(dimension)

dimensions = (1920, 1080)
print("\nUpdated dimensions:")
for dimension in dimensions:
    print(dimension)
````
