## **What is string in python?**

In python, a string is a sequence of characters.
the charcters can be:

- **letters; a, b, ..., z**
- **numbers; 1, 2, 3, e.t.c**
- **symbols; &, $, @ , # etc**
- **spaces**

Strings are **enclosed in quotes**: single('....'), double("...."), or triple('''.....'''/ """......") for multiple line

### ` Note`
String are immutable, meaning you cannot change a character directly aftre creation.

s = "Python"
s[0] = 'j' will give you an error

## **Why strings are important in Python**

Strings are everywhere in programming especially in **Engineering, Data Science and Web applications**. Strings can be used in the following
- storing and manipulating textual data like names, addresses, messages
- Handling file paths, urls and logs
- processing data from sensors, CSV files or reports
- Preparing inputs for ML models (like tokenization in NLP)

### **Why it's important to understand string methods**
Python provides many built-in string methods to manipulate and process strings efficiently. 
Knowing these methods saves time and avoid writing complex code manually.

### **COMMON EXAMPLES**

| Method | What it does | Example| Category|
|--------|--------------|--------|----------|
|**.upper()** | Convert to uppercase | "abc".upper()| Conversion Methods|
|**.lower()**| convert to lowercase| "ABC".lower() | Conversion Methods |
|**.title()**| capitalizes the first letter of each word| "python programming".title()| Conversion Methods
|**.capitalize()**| capitalize only the first character| "python".capitalize()| Conversion Methods
|**.strip()** | Removes spaces at the ends| "Hello ".strip()| Whitespace handling Methods
|**.lstrip()**| Removes whitespace from the left| "  python".lstrip()| Whitespace handling Methods
|**.rstrip()**| Removes whitespace from the right| "python ".rstrip()| Whitespace handling Methods
|**.replace()**| Replaces substring| "abc".replace("a", "x") this will give "xbc" |Replacing and modifying methods
|**.split()**| Splits string into list| "abc".split(",")  gives ["a", "b", "c"]| Replacing and modifying methods
|**.join()**| joins list into string|"-".join(["a", "b"]) gives "a-b"| Replacing and modifying methods
|**.find()**| Returns index of substring| "hello".find("e) gives will give **1**| Searching and Counting Methods
|**.index()**| returns index of substring, but if not found it raises an error| "python".index("p")| Searching and Counting Methods
|**.count()**| Counts occurrences of a substring.| "banana".count("a") | Searching and Counting Methods
|**.startswith()/ .endswith()**| Check start/end| "hello".startwith("h") gives **True**| String Checking Methods
|**.isalpha()**| checks if all characters are letters| print("python".isalpha())| String Checking Methods
|**.isdigit()**| Checks if all characters are digits| print("123".isdigit()) | String Checking Methods
|**.isalnum()**| checks if strings contains letters and digits only| print("python3".isalnum())| String Checking Methods
