# **INTRODUCTION TO FUNCTIONS**

### **`Definition`**
A function in python is a reusable blocck of code that performs a specific task. instead of writing the same again and again, you can define a function once and call it whenever you need it.

**Basic Idea** <br> 
- You give it input (optional)
- It processes it
- and gives you output (optional) 

### `Defining a Functions`
Here is a simple function that greets a user.

````python
def greet_user():
    """This function greets the user."""
    print("Hello, user!")
````

**Question** <br>
How do you call a function? <br>
**Answer** <br>
if the function does not take any input, then just write the name of the function with empty parentheses. <br>

````python
greet_user()
````

### `Passing Information to a Function`
We can modify this function to greet a user by name. you can do so by providing parameter during the function defintion.

for example 
``` python
def greet_user(username):
    """This function greet user by his username"""
    print("hello", username.title())
```



````python
def greet_user(name):
    """This function greets the user by name."""
    print(f"Hello, {name.title()}!")

greet_user("danhaya")

````
