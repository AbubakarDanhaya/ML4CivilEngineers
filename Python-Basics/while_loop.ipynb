# **WHILE LOOP**

### **`Background`**
The `for loop` takes a collection of items and executes a block of code oncefor each item in the collection. In contrast, the `while loop` runs as long as, or `while`, a certain condition is true.

### **`The while Loop in Action`**
You can use a `while` loop to count up through a series of numbers. For example, the following while loop counts from 1 to 5:

```python
current_number = 1
while current_number <= 5:
    print(current_number)
    current_number +=1

output 
1
2
3
4
5

```

````python
balance_list =[]
balance = 100
while balance >= 0:
    deposit = 25
    balance -= deposit
    balance_list.append(balance)
    

for i, d in enumerate(balance_list):
    print(f"{i+1}. {d}")


````

### **`Letting the User Close When to Quit`**

when can write a program that continuesly run until the user decided to close it.

````python
prompt = "\nTell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program."

message =""
while message != 'quit':
    message = input(prompt)
    print(message)
````

````python
prompt = "\nWhat's the headloss in the pipe?"

head_loss = int(input(prompt))
while head_loss > 1.5:
    control_loss = 0.85
    head_loss = head_loss - control_loss
    print(f"{head_loss:.2f}")
````

### **`Using a Flag`**
In the previous example, we had the program perform certain tasks while a given condition was true. But what about more complicated programs in which many different event could cause the program to stop running? <br>
for a program that shpuld run only as long as many conditiond are true, you can define one veriable that determines whether or not the entire 

````python
prompt = "\nTell me something, and I will repeat it back to you: "
prompt += "\nEnter 'quit' to end the program"

active = True
while active:
    message = input(prompt)

    if message == 'quit':
        active = False
    else:
        print(message)
````

````python
prompt = "\nWhat's your GPA? "
prompt += "\nMake sure you provide a valid GPA (0.0 - 5.0): "

active = True

while active:
    try:
        gpa = float(input(prompt))

        if 0.0 <= gpa <= 5.0:
            if gpa >= 4.50:
                print(f"You're a First Class with a GPA of {gpa}. Congratulations!!")
            elif gpa >= 3.50:
                print(f"You're a Second Class Upper with a GPA of {gpa}.")
            elif gpa >= 2.50:
                print(f"You're a Second Class Lower with a GPA of {gpa}.")
            else:
                print(f"You're a Third Class with a GPA of {gpa}.")
            
            active = False   # stop loop after valid input
        else:
            print("This is not a valid GPA. Please enter between 0.0 and 5.0.")

    except ValueError:
        print("Invalid input! Please enter a numeric GPA.")
````

### **`Using break to Exit a Loop`**
To exit a while loop immediately without running any remaining code in the loop, regardless of the results of any conditional test, use the break statement. The break statement directs the flow of your program; you can use it to control which line of code are executed and which aren't so the program only executes codes that you want it to, when you want it to.  

````python
prompt = "\nPlease enter the name of a city you have visited: "
prompt += "\n(Enter 'quit' when you are finished.) "

while True:
    city = input(prompt)

    if city.lower() == 'quit':
        break
    else:
        print(f"I've been to {city.title()} too!")
````

### **`Using continue in a Loop`**

Rather than breaking out of a loop entirely without executing the rest of its code, you can use the `continue` statement to return to the beginning of the loop, based on the result of a conditional test. 

````python
my_list =[]

current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        my_list.append(current_number)
        continue

    print(current_number)
print("This list containing even numbers", my_list)
````
