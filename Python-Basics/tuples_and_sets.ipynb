### `Defining a Tuple`
 A tuple looks just like a list, except you use parentheses instead of square brackets. Once you define a tuple, you can access individual elements by using each item's index, just as you would for a list. <br> **for example**<br> If we have a rectangle that should always be a certain size, we can ensure that its size doesn't change by putting dimensions into a tuple:

 ```
 dimensions = (200, 50)
 print("The length of the rectangle is : ", dimensions[0])
 print("The width of the rectangle is : ", dimensions[1])

 ```
