### `Looping`
Looping allows you to take the same action, or set of actions, with every item in list. as a result, you will be able to work efficently with the lists of any length, including those with thousand or millions items.

**TYPES OF LOOP**
- for loop and
- while loop

#### `for loop`
A for loop is a control structure in programming that lets you repeat a block of code a specific number of times or iterate over a sequence of items (like numbers, lists, or strings).

In simple terms
```
A for loop says:

“For each item in this collection (or for each number in this range), do this.”
```

````python
# let me print individual fruit from a fruit list
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
for fruit in fruits:
    print(fruit)
````

````python
artists = ["wizkid", "davido", "tiwa savage", "ado gwanja", "rarara", "alan waka"]
for artist in artists:
    print(artist.replace(" ", "_").strip().title())
````

## Doing something after a for loop
What happens once a for has finished executig? Usually, you'll want to summarize a block of output or move on to other work that your program must accomplished.

Any lines of code after the for loop that are not indented are excuted once without repetition.

````python
animals = ["ram", "goat", "sheep", "camel"]
for animal in animals:
    print(f"This is a {animal.title()}")
print("All of the above are example of domestic animals")
````

````python
for i, animal in enumerate(animals):
    print(f"{i + 1}. {animal.title()}")
print("We have four domestic animals listed above.")
````

### Avoid Indentation Errors
Python uses identation to determine how a line, or group of line, is related to the rest of the program. Python's use of indentaton makes code very easy to read. Basically, it uses whitespace to force you to write neatly formatted code with a clear visual structure. 

**Let's examine some common indentation error**

###  **1. Forgetting to Indent**
  Always indent the line after the for statement in a loop. If you forget, python will remain you:


````python
fruits = ["mango", "orange", "grape", "kiwi"]
for fruit in fruits:
    print(f"I like to eat {fruit.title()}")
````

````python
# without indentation
fruits = ["mango", "orange", "grape", "kiwi"]
for fruit in fruits:
print(f"I like to eat {fruit.title()}")
````

**(IndentationError: expected an indented block after 'for' statement on line 3)**

**_note: here python is given you a hint about the type bug you've and the way forward to correct it_** 

### **2. Forgetting to Indent additional lines**

Sometimes your loop will run without any errors (bugs) but won't produce the expected result. this can happen when you're trying to do several tasks in a loop and you forget to indent some of its lines.

for example, this is what happens when we forget to indent the second line in a loop that tells each fruit i would l like to eat it.

````python
fruits = ["mango", "orange", "grape", "kiwi"]
for fruit in fruits:
    print(fruit.title())
    print(f"I like to eat {fruit.title()}")
````

````python
# without indentation
fruits = ["mango", "orange", "grape", "kiwi"]
for fruit in fruits:
    print(fruit.title())
print(f"I like to eat {fruit.title()}")
````
