# **Introduction to Python Dictionary**

### **`Definition`**
A `Python` dictionary is a build-in data structure that stores data as a collection of `key-value` pairs. it is mutable, ordered and does not allow duplicate keys.

### **`Key Characteristics`**
- **`Key-Values Pairs`**: Data is stored in the format `key: value`, enclosed in curly braces `{}`.
- **`Unique and immutable keys`**: Each key within a dictionary must be unique and of an immutable data type, such as strings, numbers, or tuples.
- **`Mutable values`**: Values can be of any data type and can be duplicated across different keys.
- **`Efficient lookups`**: Dictionaries are implemented as hash tables, allowing for very fast (average O(1) time complexity) retrieval of values using their keys, which is much faster than searching in a list.

```python
my_dict = {'name': 'abubakar', 'school': 'kust wudil', 'dept': 'civil engineering', 'age': 28}
```

```python
print(my_dict)
```

**Note:** _**You can use key to access the value associated with it**_

```python
message = f"Hello, my name is {my_dict["name"].title()} and I'm a graduate of {my_dict["dept"]} from {my_dict["school"]}"
print(message)
```

```python
alien_0 = {
    "color": "green",
    "points": 5
}
print(alien_0["color"])
print(alien_0["points"])
```

**As with most new programming concepts, using dictionaries takes practice. once you've worked with dictionaries for a bit, you'll see how effectively they can model real world situations.**

### **`Accessing Values in a Dictionary`**
*To get the value associated with a key, give the name of the dictionary and then place the key inside a set of square brackets*. 

```python
student_details = {
    "first_name": "abubakar",
    "last_name": "musa",
    "registration_number": "KUST/ENG/CE/19/1234",
    "age": 28,
}
```

```python
print(f"The student's first name is {student_details['first_name'].title()} and last name is {student_details['last_name'].title()}.")
```

```python
print(f"The student's registration number is {student_details['registration_number']} and age is {student_details['age']}.")
```

### **`Adding New Key-Value Pairs`**
Dictionaries are dynamic structure, and you can add new key-value pairs to dictionaries at any time. To add a new `key-value` pair, you would give the name of the dictionary followed by the **new key** in a square brackets, along with the new value. 

```python
student_details["GPA"] = 4.0
```

```python
print(student_details)
```
