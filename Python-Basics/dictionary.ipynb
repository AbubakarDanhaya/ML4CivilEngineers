# **Introduction to Python Dictionary**

### **`Definition`**
A `Python` dictionary is a build-in data structure that stores data as a collection of `key-value` pairs. it is mutable, ordered and does not allow duplicate keys.

### **`Key Characteristics`**
- **`Key-Values Pairs`**: Data is stored in the format `key: value`, enclosed in curly braces `{}`.
- **`Unique and immutable keys`**: Each key within a dictionary must be unique and of an immutable data type, such as strings, numbers, or tuples.
- **`Mutable values`**: Values can be of any data type and can be duplicated across different keys.
- **`Efficient lookups`**: Dictionaries are implemented as hash tables, allowing for very fast (average O(1) time complexity) retrieval of values using their keys, which is much faster than searching in a list.

````python
my_dict = {'name': 'abubakar', 'school': 'kust wudil', 'dept': 'civil engineering', 'age': 28}
````

````python
print(my_dict)
````

**Note:** _**You can use key to access the value associated with it**_

````python
message = f"Hello, my name is {my_dict["name"].title()} and I'm a graduate of {my_dict["dept"]} from {my_dict["school"]}"
print(message)
````

````python
alien_0 = {
    "color": "green",
    "points": 5
}
print(alien_0["color"])
print(alien_0["points"])
````

**As with most new programming concepts, using dictionaries takes practice. once you've worked with dictionaries for a bit, you'll see how effectively they can model real world situations.**

### **`Accessing Values in a Dictionary`**
*To get the value associated with a key, give the name of the dictionary and then place the key inside a set of square brackets*. 

````python
student_details = {
    "first_name": "abubakar",
    "last_name": "musa",
    "registration_number": "KUST/ENG/CE/19/1234",
    "age": 28,
}
````

````python
print(f"The student's first name is {student_details['first_name'].title()} and last name is {student_details['last_name'].title()}.")
````

````python
print(f"The student's registration number is {student_details['registration_number']} and age is {student_details['age']}.")
````

### **`Adding New Key-Value Pairs`**
Dictionaries are dynamic structure, and you can add new key-value pairs to dictionaries at any time. To add a new `key-value` pair, you would give the name of the dictionary followed by the **new key** in a square brackets, along with the new value. 

````python
student_details["GPA"] = 4.0
````

````python
print(student_details)
````

### **`Starting with an Empty Dictionary`**
It's sometimes convenient, or neccessary, to start with an empty dictionary and then add each new item to it. To start filling an empty dictionary, define a dictionary with an empty set of braces and then add each `key-value` pair on its own line.

```python

my_details = {}
my_details["first_name"]= "abubakar"
my_details["last_name"]= "danhaya"
print(my_details) ### output {"first_name": "abubakar", "last_name": "danhaya"}

```


### **`Modifying Values in a Dictionary`**
To modify a value in a dictionary, give the name of the dictionary with the key in a square brackets and then the new value you want associated with that key.

```python
my_details["first_name"] = "sadiq"
print(my_details) ### output {"first_name": "sadiq", "last_name": "danhaya"}

````python
student_details = {
    "first_name": "abubakar",
    "last_name": "musa",
    "registration_number": "KUST/ENG/CE/19/1234",
    "age": 28,
    "level": 1,
    "min_yrs_to_graduate": 2
}
if student_details["min_yrs_to_graduate"]  < 4:
    student_details["min_yrs_to_graduate"] = 4
if student_details["level"] == 1:
    level_status = "freshman"
elif student_details["level"] == 2:
    level_status = "sophomore"
elif student_details["level"] == 3:
    level_status = "junior"
else:
    level_status = "senior"
print(f"The student is a {level_status}.")
print(student_details)


````

### **`Removing key-value pairs`**

When you no longer need a piece of information that's stored in a dictionary, you can use the `del` statement to completely remove a `key-value` pair. All `del` needs the name of the dictionary and the key that you want to remove.

````python
my_details = {
    "first_name": "abubakar",
    "email": "habudanhayabk@gmail.com",
    "phone_number": "08012345678",
    "major": "civil engineering",
    "highest_education": "bachelor's degree",
}

if my_details['email'] == "habudanhayabk@gmail.com":
    my_details['alternative_email'] = "abubakardanhaya@outlook.com"
if my_details["email"] == my_details["alternative_email"]:
    del my_details["alternative_email"]
print(my_details)
````

### **`Using get() to Access values`**
Using keys in square brackets to retrieve the value you're interested in from a dictionary might cause one potential problem: if the key you ask for doesn't exist, you'll get an error. <br>
The `get()` method requires a `key` as a first argument. As a second optional argument, you can pass the pass the value to be returned if the key doesn't exist.

````python
science_subjects = {
    "physics": "the study of matter and energy",
    "chemistry": "the study of substances and their properties",
    "biology": "the study of living organisms",
    "mathematics": "the study of numbers, quantities, and shapes",
    "computer science": "the study of computers and computational systems",
}
print(f"The definition of physics is: {science_subjects.get('physics')}")
````

````python
print(science_subjects.get('astronomy', 'The subject astronomy is not found in the dictionary.'))
````

````python
alien_0 = {
    "color": "green",
    "points": 5,
}
print(alien_0.get("speed", "The key 'speed' is not found in the dictionary."))
````

## **`Looping Through a Dictionary`**
A single python dictionary can contain just few `key-value` pairs or millions of pairs. Because a dictionary can contain large amounts of data, Python lets you loop through a dictionary. dictionaries can be used to store information in a variety of ways; therefore several ways exist to loop through them. You can loop through all of a dictionary's `key-value` pairs, through its keys, or through its values.


### **`Looping Through All Key-Value Pairs`**
`.items()` is used to access both keys and values in a dictionary.

````python
best_football_team = {
    "team_name": "Manchester United",
    "manager": "Micheal Carrick",
    "stadium": "Old Trafford",
    "league": "Premier League",
    "jersey_colors": ["red", "white", "black"],
    "number_of_premier_league_titles": 20,
    "number_of_champions_league_titles": 3,
    "number_of_fa_cup_titles": 12,
    
}
print("Ladies and gentlemen, the best football team in the world is:")
for key, value in best_football_team.items():
    print(f"{key.title()}: {value}")
````
