# **CONDITIONAL STATEMENTS IN PYTHON**

### **`Definations`**
**Conditional statements in python are used to execute specific blocks of code based on whether a given conditions is `True` or `False`, allowing the program to make decisions and control its flow. The primary keywords used are `if`, `elif` and `else`.**

### **`Core Components`**
- **`if` statement** <br>
- **`elif` statement** and <br>
- **`else` statement**
  
_Programming often involves examining a set of conditions and deciding which action to take based on those conditions. `Python's` conditional statement allows you to examine the current state of a program and respond appropriately to that state._

#### **`If` statement**
The simplest form, which runs a block of code only if its condition evaluates to `True`.

```python
# syntax
if condition:
    this part of code runs for truthy conditions
```

````python
age = 19
if age >= 18:
    print("You are old enough to votes!")
    print("Have you registered to vote yet?")
````

````python
age = 20
if age >= 18:
    print("You are an adult.")
````

sometimes you may require user input

````python
age = int(input("pls write you age"))
if age > 20:
    print("You're in high school")
````
