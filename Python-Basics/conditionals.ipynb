### **`If-else` Statement**

Often, you will want to take one action when a conditional test passes and a different action in all other cases. `Python's` `if-else` syntax makes this possible. An `if-else` block is similar to a simple `if` statement, but the `else` statement allows you to define an action or set of actions that are executed when the conditional test fails. 

````python
age = int(input("pls write you age"))
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote yet.")
````

````python
slab_thickness = float(input("Enter slab thickness in meters: "))
if slab_thickness > 0.15:
    print("Check deflection criteria.")
else:
    print("Deflection criteria is satisfactory.")
````

### **`if-elif-else` Statement**
Often, you'll need to test more than two possible situations, and to evaluate these you can use `python` `if-elif-else` syntax. Python executes only one block in an `if-elif-else` chain. it runs each conditional test in order, until one passes. When a test passes, the code following that test is executed and `Python` skips the rest of the tests.

let's consider this admission process

- Admission for anyone under age 4 is free
- Admission for anyone between the age of 4 and 18 is 25,000 Naira
- Admission for anyone age 18 or older is 40,000 Naira

````python
gpa = float(input("Enter your GPA: "))
if gpa >= 4.50:
    print("You are a first class student.")
elif gpa >= 3.50:
    print("You are a second class upper student.")
elif gpa >= 2.50:
    print("You are a second class lower student.")
elif gpa >= 1.50:
    print("You are a third class student.")
else:
    print("You have failed. Better luck next time.")
````

````python
age = int(input("What's your age?"))
if age < 4:
    print("You are admission cost is free!")
elif 4 < age < 18:
    print("Your admission cost is 25,000 Naira")
else:
    print("Your admission cost is 40,000")
````

````python
age = int(input("what's your age?"))
if age < 4 :
    price = 0
elif age < 18:
    price = 25_000
elif age < 65:
    price = 40_000
else:
    price = 20_000

print(f"Your admission cost is {price} Naira.")
````

### **`Testing Multiple Conditions`**
The `if-elif-else` chain is powerful, but it's only appropriate to use when you just need one test to pass. As soon as `Python` finds one test that passes, it skips the rest of the tests. This behavior is beneficial, because it's efficient and allows you to test for one specific condition. <br>
However, sometimes it's important to check all conditions of interest. In this case, you should use a series of simple `if` statement with no `elif or else` blocks. This technique makes sense when more than one condition could be `True`, and you want act on every condition that is `True`.

*let's consider the pizzeria example. If someone requests a two-topping pizza, you'll need to be sure to include both toppings on their pizza.*

````python
requested_ingredients = ['cement', 'sand', 'gravel']
if 'cement' in requested_ingredients:
    print("Pls i need cement")
if 'steel' in requested_ingredients:
    print("Pls i need steel")
if 'sand' in requested_ingredients:
    print("Pls i need sand")
````

````python
requested_toppings = ['mushrooms', 'extra cheese', 'green peppers']
if 'mushrooms' in requested_toppings:
    print("Adding mushrooms.")
if 'pepperoni' in requested_toppings:
    print("Adding pepperoni.")
if 'extra cheese' in requested_toppings:
    print("Adding extra cheese.")
print("\nFinished making your pizza!")
````

This code would not work properly if we used an `if-elif-else` block, because the code would stop running after only one test passes. Here's what that would look like:

````python
requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']
if 'mushrooms' in requested_toppings:
    print("Adding mushrooms.")
elif 'pepperoni' in requested_toppings:
    print("Adding pepperoni.")
elif 'extra cheese' in requested_toppings:
    print("Adding extra cheese.")

print("\nFinished making your pizza!")
````

_In summary, if you want only one block of code to run use `if-elif-else` chain. If more than one block of code needs to run, use a series of of independent `if` statements._

### **Using `if` statement with Lists**
`if` statement can be used for many operations with lists for examples:
-  Checking for special items
-  Checking that a list is not empty

````python
attendance = ["bash", "anas", "ali", "zarewa", "zakiyyu", "abba", "shinga"]
for name in attendance:
    print(f"Sir, i am here {name.title()}")
````

````python
students = ["bash", "anas", "ali", "zarewa", "zakiyyu", "abba", "shinga", "namzee", "makole", "observer"]
attendance = ["bash", "anas", "ali", "zarewa", "zakiyyu", "abba", "shinga"]
my_list = []
for name in students:
    if name in attendance:
        my_list.append(name)
print("Names of those who attended the class: ")
for name in my_list:
    print(name.title())
````

````python
requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']
for requested_topping in requested_toppings:
    print(f"Adding {requested_topping}.")
print("\nFinished making your pizza!")
````

````python
exhausted_toppings = ['green peppers', 'pepperoni']
for requested_topping in requested_toppings:
    if requested_topping in exhausted_toppings:
        print(f"Sorry, we are out of {requested_topping} right now.")
    else:
        print(f"Adding {requested_topping}.")
````

````python
concrete_ingredients = ['cement', 'sand', 'gravel', 'water', 'admixture']
requested_ingredients = ['cement', 'sand', 'steel', 'water']
````

````python
for requested_ingredient in requested_ingredients:
    if requested_ingredient in concrete_ingredients:
        print(f"Adding {requested_ingredient}.")
    else:
        print(f"Sorry, we don't have {requested_ingredient}.")
````

````python
available_toppings = ['mushrooms', 'olives', 'green peppers',
                      'pepperoni', 'pineapple', 'extra cheese']
requested_toppings = ['mushrooms', 'french fries', 'extra cheese']

for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print(f"Adding {requested_topping}.")
    else:
        print(f"Sorry, we don't have {requested_topping}.")

print("\nFinished making your pizza!")
````
